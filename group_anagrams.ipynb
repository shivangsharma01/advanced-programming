# Leetcode 49 Group anagrams

```python
# What if i make a key for every string by sorting it and then based on that key store all the possible anagrams occuring in the list

# Hence this will be a list of dict OR defaultdict(list) - WHY - defaultdict helps us initialize a null entry on its own we can handle that ourselves also

from collections import defaultdict
from typing import List

class Solution:
    def group_anagrams(self, strs: List[str])->List[List[str]]:

        groups = defaultdict()

        for word in strs:

            key = "".join(sorted(word))

            groups[key].append(word)

        return list(groups.values())
```

## Can you code it without defaultdict()

```python
class Solution:
    def group_anagrams(self, strs: List[str])->List[List[str]]:

        groups = {}

        for word in strs:
            key = "".join(sorted(word))

            if key not in groups:
                groups[key]=[]
            groups[key].append(word)

        return list(groups.values())

# TC: O(n * klog(k))
# SC: O(n * k)
```

## Solve without using sorted()

## Instead of sorted() - create a character frequency mapping - create a tuple for that

```python
class Solution:
    def group_anagrams(self, strs: List[str])->List[List[str]]:

        groups = {}

        for word in strs:

            count = [0]*26

            for ch in word:

                count[ord(ch)-ord('a')]+=1 

            key = tuple(count)

            if key not in groups:

                groups[key] = []

            groups[key].append(word)

        return list(groups.values())

# k - average number of charachters in a word
#  TC: O(N * k * 26) = O(N*k)
#  SC: tuple O(26) = O(1) hashmap: O(n*k) 
```

### Why defaultdict() - whenever i access a missing key - it automatically creates an empty list for me

#### Better explaination

```python
from typing import List

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:

        groups = {}

        for word in strs:

            count = [0]*26 # creating a character frequency mapping

            # shivang: ['a':1, 'b':0, 'c': 0,........'g':1, 'h':1, 'i':1,......]

            for ch in word:

                count[ord(ch)-ord('a')]+=1
                # increase the frequency of the character byt getting the absolute position using ord()

            # list is mutable - tuple is immutable
            # so list cannot be hashed - whereas we are using a hashmap which creates a hash table - tuple can be hashed
            
            key = tuple(count)

            if key not in groups:

                groups[key] = []
            
            groups[key].append(word)

        return list(groups.values())
```

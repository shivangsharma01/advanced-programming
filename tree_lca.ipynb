# Leetcode 236

## Finding the least common ancestor

## Think of it as a postorder DFS problem

## LEFT RIGHT ROOT

```python
#         3
#       /   \
#      5     1
#     / \
#    6   2
#       / \
#      7   4

# p = 6
# q = 4

# The answer is 5
```

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.right = None
        self.left = None

class Solution:
    def lowest_common_ancestor(self, p:TreeNode, q:TreeNode, root: TreeNode)-> TreeNode:

        # POSTORDER DFS: LEFT RIGHT ROOT

        def dfs(node):

            # If there is an empty node - simply return none
            if node is None:
                return None

            # found p or q? - return that node - why? because that is my goal - if i find p or q- return the node
            if p==node or q==node:
                return node
            
            # recursively iterate through left 
            left = dfs(node.left)
            # then right
            right = dfs(node.right)

            # if left and right both found p or q respectively then this is the lowest node
            if left and right:
                return node
            
            # if both were not found - return the left one first 
            if left:
                return left

            # if left also not found - return right
            return right
```

## I am given a number - i have to return if it is prime or not

### Prime - divisble only by 1 and itself

### I will divide it all the numbers ranging from 2 to n-1 - if it is divisible i.e. the remainder is 0 - return false

```python
class Solution:
    def is_prime(self, n:int):
        isprime = True

        if n == 1:
            isprime = False

        # Example the input is 100
        for i in range(2,n-1):

            if n%i==0:
                isprime = False
        
        return isprime
    
    def print_primes(self, x:int):

        for i in range(1, x+1):

            print(f'{i} \t',self.is_prime(i))

obj = Solution()
obj.print_primes(20)
```

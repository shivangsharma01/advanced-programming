# Leetcode 752

Can I describe the problem as moving between states?
                │
               Yes
                │
Does each move have the same cost?
                │
               Yes
                │
Am I looking for the minimum number of moves?
                │
               Yes
                │
              → BFS

That's the intuition interviewers are looking for. They don't want you to recognize "Open Lock"; they want you to recognize the pattern: an implicit graph, uniform edge costs, and a shortest-path objective. Once you see those three ingredients together, BFS becomes the natural choice rather than a memorized solution.

```python
from collections import deque

class Solution:
    def open_lock(self, deadends:list[str], target: str)-> int:

        dead = set(deadends) # i will create a set of deadends not because i think there will be duplicates in the given list of str but because set uses hashtable and has O(1) lookup

        if "0000" in dead: # O(1) lookup using set() as it uses a hashtable
            return -1
        
        if target == "0000":
            return 0
        
        queue = deque([("0000","0")])  # Queue stores the state and moves in a list where they are coupled together as (state, moves)
        
        # why deque - this solution creates an implicit graph which where cost of every move is same & i need to find the minimum no of moves - so it is a BFS problem
        # BFS follows FIFO First in First out
        # Python list costs me O(N) time for front removals - collections.deque() gives me O(1) front removal using popleft()


        visited = {0000} # this is a set - no key - value pairs - only value
        
        # dict() creates a dictionary but it will have key and value pairs
        # set() can be used but set() gives me an empty set - if i want an initialised set i have to declare it like: {"1000"} with 1000 as the first value
        
        while queue:
            # keep popping the nodes from the front and adding from the back

            node, moves = queue.popleft() 
            # this gives me nodes = "0000" and moves = "0"

            if node == target:
                return moves

            chars = list(node)

            for i in range(4):

                digit = int(chars[i])

                # forward first
                chars[i] = str((digit+1)%10)
                neighbor = "".join(chars)

                if neighbor not in visited and neighbor not in dead:
                    queue.append((neighbor, moves+1))
                    visited.add(neighbor)
                
                # backward shift
                chars[i] = str((digit-1)%10)
                neighbor = "".join(chars)

                if neighbor not in visited and neighbor not in dead:
                    queue.append((neighbor,moves+1))
                    visited.add(neighbor)
                
                # restore the original character - so that we can iterate unknown nodes
                # MUST RESTORE THE DIGIT IN STRING FORMAT
                chars[i] = str(digit)

        return -1
```

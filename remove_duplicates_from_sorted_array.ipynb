```python
class Solution:
    def remove_duplicates(self, nums:list[int])->int:

        # handling basic usecase
        if not nums:
            return
        
        # this will track the last element with unique element
        i = 0

        # j iterates through the all the elements
        for j in range(len(nums)):

            # if it spots another unique element
            if nums[i] != nums[j]:

                i+=1
                nums[i]=nums[j]
            
        return i+1
```

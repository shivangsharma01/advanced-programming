# Leetcode 26

#### "Since the array is sorted, duplicate values are adjacent. I'll use two pointers. j scans through the array, while i tracks the position of the last unique element. Whenever nums[j] differs from nums[i], I've found a new unique value, so I increment i and write that value into nums[i]. At the end, i + 1 gives the number of unique elements."

```python
class Solution:
    def remove_duplicates(self, nums:list[int])->int:

        # handling basic usecase
        if not nums:
            return
        
        # this will track the last element with unique element
        i = 0

        # j iterates through the all the elements
        for j in range(len(nums)):

            # if it spots another unique element
            if nums[i] != nums[j]:

                i+=1
                nums[i]=nums[j]
            
        return i+1
```

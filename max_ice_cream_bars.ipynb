# Leetcode 1833

## w/o counting sort

```python
class Solution:
    def maxIceCream(self, costs: list[int], coins: int) -> int:

        # I will apply greedy logic here - first i will buy the cheapest bars and then move on to the expensive bars
        costs.sort()

        bars = 0

        for cost in costs:

            if coins>=cost:
                coins-=cost
                bars+=1
            else:
                break
        return bars

# TC: sorting O(NlogN) and traversal: O(N) - Total TC: O(N)
# SC: O(1)
```

## With counting sort

```python
class Solution:
    def maxIceCream(self, costs: list[int], coins: int) -> int:
        
        freq = [0]*100001

        for cost in costs:
            freq[cost]+=1

        bars = 0

        for cost in range(0,100001):

            if freq[cost]==0:
                continue
            
            buy = min(freq[cost], coins//cost)

            coins -= buy*cost

            bars+=buy

            if coins < cost:
                break
        return bars
# TC : O(N+MAXCOST) - MAXCOST HERE IS 100001
# SC: O(100001)
```

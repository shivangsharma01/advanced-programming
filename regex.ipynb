# Leetcode 10

### Regular Expression Matching

### DP: Recursion + Memoization

### when should we stop - when the pattern p has been completed or len(p) == j or the current pointer iterating p

### what next? - i will check if there is a * or a . - '*' will make the problem more complex

```python
from functools import lru_cache

class Solution:
    def isMatch(self, s: str, p: str) -> bool:

        @lru_cache(None)
        def dp(i,j):

            if j == len(p):
                return i == len(s)

            first_match = (
                
                i<len(s) 

                and

                (
                    s[i] == p[j] or p[j]=='.'
                )
            )

            if j+1 < len(p) and p[j+1]=='*':

                # now i am taking into account x*

                # consume or skip

                skip_star = dp(i,j+2)

                consume_star = first_match and dp(i+1,j)

                return skip_star or consume_star

            return first_match and dp(i+1,j+1)
        
        return dp(0,0)

# TC: O(m*n) where m is length of string and n is string of pattern
# SC: O(m*n) because of memoization
```

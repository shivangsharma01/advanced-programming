# Rotate array Leetcode 189

```python
class Solution:

    def rotate_array(self, nums:list[int], k:int)-> None:

        for _ in range(k):

            last = nums[-1]

            # this loop will stop before reaching '0'th index
            for i in range(len(nums)-1,0,-1):
                nums[i] = nums[i-1]
            
            nums[0] = last

# TC: O(n*k)
# SC: O(1)
# This code will hit TLE
```

# Optimised - O(n)

```python
class Solution:

    def rotate_array(self, nums:list[int], k:int)-> None:

        res = [0]*len(nums)

        for i in range(nums):

            res[(i+k)%len(nums)] = nums[i] # k%n because lets say k is bigger than n - e.g. k = 10 n = 7 - here i can simply rotate 3 times to get the resultant array

        for i in range(nums):

            nums[i] = res[i]

# TC: O(n)
# SC: O(n)
```

# Can you do it inplace

### Rotation means - reverse the complete array - then reverse the array from 0 to k - then reverse the array from k to n-1
### Then we get the answer in place

```python
class Solution:

    def rotate_array(self, nums:list[int], k:int)-> None:

        def rotate(left, right):

            while left<right:

                nums[left], nums[right] = nums[right], nums[left]
                left+=1
                right-=1

            return nums

        n = len(nums)
        k = k%n

        nums = rotate(0,n-1)

        nums = rotate(k,n-1)

        nums = rotate(0,k-1)

# TC: O(N) + O(K)+ O(N-K) = O(N)
# SC : O(1)
```

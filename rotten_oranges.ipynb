## Rotten Oranges

```python
from collections import deque

class Solution:
    def rotten_oranges(self, grid:list[list[int]])-> int:

        rows = len(grid)
        cols = len(grid[0])

        # this is a multi source BFS problem - constant cost of traversal
        # why not use DFS - because it will go to the depth of that tree and then return
        # but we want one step at a time - similar to waves
        # maintain a queue which stores all the fresh oranges then explore their neighbors iteratively

        queue = deque()

        fresh = 0

        for r in rows:
            for c in cols:

                if grid[r][c]==2:

                    queue.append(r,c)
                
                if grid[r][c]==1:

                    fresh+=1

        # now that i have the fresh orange count and queue for iteration
        # i need direction for each node traversal

        directions = [
            (1,0), # top
            (-1,0), # bottom
            (0,1), # right
            (0,-1) # left
        ]

        minutes = 0

        while queue:

            # Iterating through rotten oranges one step at a time in the queue and rotting neighboring oranges
            for _ in range(len(queue)):

                # one wave = one BFS step

                r,c = queue.popleft()

                for dr,dc in directions:
                    
                    nr = r + dr
                    nc = c + dc

                    if 0<=nr<rows and 0<=nc<cols and grid[nr][nc]==1:

                        grid[nr][nc]=2
                        fresh-=1
                        queue.append(nr,nc)

                if queue:
                    minutes+=1
                
        return minutes if fresh==0 else -1
```

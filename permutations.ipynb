# Leetcode 46

### Think of it as DFS

```python
# Finding the permutation takes N factorial OR N!

class Solution:
    def permute(self, nums: list[int]) -> list[list[int]]:

        visited = set()

        res = []

        current = []

        def backtrack():

            if len(current)==len(nums):
                res.append(current.copy())
                return
            
            for num in nums:

                if num in visited:

                    continue

                current.append(num)
                visited.add(num)

                backtrack()

                current.pop()
                visited.remove(num)

        backtrack()
        return res
```

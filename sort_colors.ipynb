```python
class Solution:
    def sortColors(self, nums: list[int]) -> None:
        # i will split the array into 3 sections and for that i need 4 boundaries

        # low mid <unknown> high

        # we will operate in the unknown boundary in b/w mid and high and keep replacing array elements in place

        low = 0
        mid = 0
        high = len(nums)-1

        while mid<=high: # goal is to shrink the unknown window

            # NOTE We will only use nums[mid] to move through the array as it has access to both the boundaries

            if nums[mid]==0:
                # we need to send it to the beginning of the array
                # swap mid with low
                nums[low], nums[mid] = nums[mid], nums[low]
                low+=1
                mid+=1
            
            elif nums[mid]==1:
                mid+=1
        
            elif nums[mid]==2:
                # swap mid with high BUT do not increase the mid pointer
                nums[mid], nums[high] = nums[high], nums[mid]
                # we will deliberatly not increase mid because it needs to be checked again after swapping

                high-=1
```

## Create nodes of the unique values found

```python
from typing import List
from collections import defaultdict
class TreeNode:
    def __init__(self, val:int, left = None, right = None)-> None:
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def createBinaryTree(self, description:List[List[int]])-> TreeNode:

        nodes = {}
        # nodes = defaultdict(TreeNode)

        # goal is to find the root node & i can easily track all the children values and return the missing value
        children = set()


        for parentval, childval, isleft in description:

            if parentval not in nodes:
                # unique parent value found

                nodes[parentval] = TreeNode(parentval)

            if childval not in nodes:
                # unique child value found

                nodes[childval] = TreeNode(childval)
            
            # Now i have to create the binary tree by joining the relevant values

            parent = nodes[parentval]
            child = nodes[childval]

            if isleft==1:
                parent.left = child
            else:
                parent.right = child
            
            # remember set() does not have append() it has add()
            children.add()

        # i will iterate through hashmap - how - items()

        for value, node in nodes.items():

            if value not in children:
                return node
```

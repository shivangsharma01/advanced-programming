```python
class Solution:
    def search(self, nums: list[int], target: int) -> int:

        # Only using binary search is not enough in this problem

        # i need to find the sorted part of the array also

        left = 0
        right = len(nums)-1

        while left<=right:

            # <= because there should be 1 element left in the search space at the end

            mid = (left+right)//2

            if nums[mid]==target:
                return mid

            # is the left array sorted?

            if nums[left]<=nums[mid]: # <= because every number should either be less than or equal to nums[mid]

                # But is target present in the left side of the array
                if nums[left]<= target<nums[mid]:
                    # target can be equal to nums[left] but must be strictly less than mid
                    right = mid-1
                else:
                    left = mid+1
            else:

                if nums[mid]<= target< nums[right]:
                    left = mid+1
                else:
                    right = mid-1

        return -1
```
